# Embeddings model exploration

```python
from sentence_transformers import SentenceTransformer

from jobby.config import settings

model_name = settings.embedding_model
model = SentenceTransformer(model_name)
```

```python
emb1 = model.encode_query("Erfahrung mit PyTorch und neuronalen Netzen")
emb2 = model.encode_query("Deep Learning Modelle trainiert")
emb3 = model.encode_query("Fußball und Basketball")
emb4 = model.encode_query("Ein spieler spielt den Ball")
```

```python
print(emb1)
print(len(emb1))
print(type(emb1))
```

```python
print(model.similarity(emb1, emb2))
print(model.similarity(emb2, emb3))
print(model.similarity(emb3, emb4))
```
