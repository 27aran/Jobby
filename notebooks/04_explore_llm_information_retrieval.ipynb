```python
response = chat(
    model="qwen3.5:4b",
    messages=[{"role": "user", "content": build_extract_prompt(job_description)}],
    format=RequirementList.model_json_schema(),
    think=False,
    options = {'temperature':0},
)

result = RequirementList.model_validate_json(response.message.content)

for req in result.requirements:
    print(f"[{req.importance:12}] [{req.category:10}] {req.text}")
```
